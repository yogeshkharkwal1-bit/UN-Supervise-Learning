```python
import seaborn as sns
from sklearn.datasets import make_blobs
```

```python
X,y=make_blobs(
    n_samples=1000,
    n_features=2,
    centers=4,
    random_state=42
)
```

```python
sns.scatterplot(x=X[:, 0],y=X[:, 1])
```

# how to choose the k-value

```python
from sklearn.cluster import KMeans
wcss=[]
for k in range(1,21):
    kmeans=KMeans(n_clusters=k)
    kmeans.fit_predict(X)
    wcss.append(kmeans.inertia_)
    
```

```python
sns.lineplot(x=range(1,21),y=wcss,marker='0')
```

## Manual way to find the best k value
1. we do not need to create any plot for visualizing and finding thw k value
2. KneeLocator it find thw knee/elboe value

```python
from kneed import KneeLocator
```

```python
from sklearn.cluster import KMeans
wcss=[]
for k in range(1,21):
    kmeans=KMeans(n_clusters=k)
    kmeans.fit_predict(X)
    wcss.append(kmeans.inertia_)
    
knee=KneeLocator(range(1,21),wcss,curve="convex",direction="decreasing")
```

```python
print("optimal l",knee.knee)
print(knee.elbow)
```
