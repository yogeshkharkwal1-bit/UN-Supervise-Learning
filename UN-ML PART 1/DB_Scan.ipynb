```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
iris=load_iris()

X=iris.data
y=iris.target
```

```python
scaling=StandardScaler()
X_scaled=scaling.fit_transform(X)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,2],marker='o')
```

```python
#DBSCAN

from sklearn.cluster import DBSCAN

DS=DBSCAN(
    eps=0.8,
    min_samples=5
)

```

```python
label=DS.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,1],marker='o',c=label)
```

# DBSCAN for  non linear datasets

```python
# make_moon create a non linear datasets 

from sklearn.datasets import make_moons

X,y=make_moons(
    n_samples=1000,
    noise=0.05,
    random_state=42
)
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,1],marker='o')
```

## Apply k-means

```python
from sklearn.cluster import KMeans

km=KMeans(
    n_clusters=2,
    random_state=42
)

labels=km.fit_predict(X_scaled)

sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,1],marker='o',c=labels)
```

# Apply DBSCAN

```python
ds=DBSCAN(
    eps=0.5,
    min_samples=5
)

label=ds.fit_predict(X_scaled)

sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,1],c=label)
```
