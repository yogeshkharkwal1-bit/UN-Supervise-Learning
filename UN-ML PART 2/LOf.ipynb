# USING THE LOCAL OURLIER FACTOR FOR OUTLIERS

```python
from sklearn.preprocessing import StandardScaler
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv("thyroid_dataset.csv")
df.tail()
```

```python
X=df.drop("Outlier_label",axis=1)
y=df["Outlier_label"]
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
from sklearn.neighbors import LocalOutlierFactor

loc=LocalOutlierFactor()
# loc=LocalOutlierFactor(contamination=0.036)
label=loc.fit_predict(X_scaled)
```

```python
from sklearn.decomposition import PCA

pca=PCA(n_components=2)
X_pca=pca.fit_transform(X_scaled)
```

```python
sns.scatterplot(x=X_pca[:,0],y=X_pca[:,1],c=label)
```

```python
import numpy as np
n_outliers= np.sum(label== -1)
n_normal= np.sum(label== 1)

print(n_outliers)
print(n_normal)
```
